```python
X = data_temp.drop(columns =['store_and_fwd_flag','dropoff_longitude','trip_duration'])
y = data_temp['trip_duration']
```

```python
X.sample()
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, f1_score, classification_report
from sklearn.model_selection import train_test_split
from sklearn. linear_model import LogisticRegression

from sklearn.svm import SVC
```

```python
X_train,X_test,y_train,y_test = train_test_split(
    X,y,random_state=42,test_size=0.20
)
```

```python
models = {
    'KNN' : KNeighborsClassifier(),
    'Navie_bayes' : GaussianNB(),
    'Decision_tree' : DecisionTreeClassifier(),
    'SVC' : SVC(),
    'Logistic Regression' : LogisticRegression(),
}
```

```python
scaler = StandardScaler()
```

```python
X_train.select_dtypes(include=["datetime64[ns]"]).columns
```

```python
for df in [X_train, X_test]:
    df["pickup_hour"] = df["pickup_datetime"].dt.hour
    df["pickup_day"] = df["pickup_datetime"].dt.day
    df["pickup_month"] = df["pickup_datetime"].dt.month
    df["pickup_weekday"] = df["pickup_datetime"].dt.dayofweek
X_train.drop(columns=["pickup_datetime", "dropoff_datetime"], inplace=True)
X_test.drop(columns=["pickup_datetime", "dropoff_datetime"], inplace=True)

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.fit_transform(X_test)
```

```python
result = []
for name,model in models.items():
    model.fit(X_train_scaled,y_train)
    y_pred = model.predict(X_test_scaled)
    acc = accuracy_score(y_test,y_pred)
    f1 = f1_score(y_test,y_pred)
    result.append({
        'model':name,
        'Accuracy':round(acc,4),
        "f1 score":round(f1,4)
        }
    )
```
