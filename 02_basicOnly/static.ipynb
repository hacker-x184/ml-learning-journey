```python
import numpy as np
import pandas as pd
```

```python

# Set seed for reproducibility
np.random.seed(42)

# Define dataset size (60 samples, falling in the 50-100 range)
n_samples = 60

# Generate synthetic features
cgpa = np.random.uniform(6.0, 10.0, n_samples).round(2)
study_hours = np.random.normal(5.0, 1.5, n_samples).clip(1.0, 10.0).round(1)
aptitude_score = np.random.randint(50, 100, n_samples)

# Generate a synthetic binary target (placement status: 0 or 1) influenced by CGPA and study hours
score_metric = (cgpa - 6.0) * 2 + (study_hours * 0.5)
probability = 1 / (1 + np.exp(-(score_metric - 4)))
placement_status = (np.random.random(n_samples) < probability).astype(int)

# Combine all arrays column-wise into a single 2D NumPy dataset
# Columns: [CGPA, Study_Hours, Aptitude_Score, Placement_Status]
dataset = np.column_stack((cgpa, study_hours, aptitude_score, placement_status))

# Display dataset details
print("Dataset Shape:", dataset.shape)
print("\nFirst 5 rows [CGPA, Study Hours, Aptitude Score, Placement Status]:")
print(dataset[:5])
df = pd.DataFrame(dataset, columns=['cgpa', 'study_hours', 'aptitude_score', 'placement_status'])

# First 5 rows check karne ke liye
print(df.head())

# DataFrame ka basic structure aur data types dekhne ke liye
print("\nDataFrame Info:")
print(df.info())
```
